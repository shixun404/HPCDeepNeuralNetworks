```python
import pandas as pd
import matplotlib.pyplot as plt
# Load the data into a pandas DataFrame
from io import StringIO

fig, ax = plt.subplots(figsize=(8, 6))
# Plot the third column (index 3) vs the fourth column (index 4)
name_list = ['cufft.csv', 'test.csv', 'test_opt.csv', 'test_opt_1.csv']
label = ['cuFFT', 'turboFFT', 'turboFFT_opt', 'turboFFT_opt_2']
for i in [0, 2]:
	with open(name_list[i], 'r') as f:
		data = f.read()
	df = pd.read_csv(StringIO(data), header=None)
	ax.plot(df[2], df[3], marker='o', label=label[i])


ax.set_xlabel('Batch Size')
ax.set_ylabel('Execution Time (ms)')
ax.set_title('256-point 1D FFT, keep 64 lower frequencies')
ax.grid(True)
ax.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
# Load the data into a pandas DataFrame
from io import StringIO

fig, ax = plt.subplots(figsize=(8, 6))
# Plot the third column (index 3) vs the fourth column (index 4)
name_list = ['cufft.csv', 'test.csv', 'test_opt.csv', 'test_opt_1.csv']
label = ['cuFFT', 'turboFFT', 'turboFFT_opt', 'turboFFT_opt_2']
for i in range(1):
	with open(name_list[i], 'r') as f:
		data = f.read()
	df = pd.read_csv(StringIO(data), header=None)
	ax.plot(df[2], df[3] * 2, marker='o', label=label[i])

with open(name_list[2], 'r') as f:
	data = f.read()
df = pd.read_csv(StringIO(data), header=None)
with open(name_list[3], 'r') as f:
	data = f.read()
df1 = pd.read_csv(StringIO(data), header=None)
ax.plot(df[2], df[3] + df1[3], marker='o', label=label[2])

ax.set_xlabel('Batch Size')
ax.set_ylabel('Execution Time (ms)')
ax.set_title('256-point 2D FFT, keep 64 lower frequencies')
ax.grid(True)
ax.legend()
plt.show()
```
